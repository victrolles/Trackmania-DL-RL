```python
import numpy as np
import cv2
import os
```

```python
interval = []       # [start, end] if empty, all files in image directory will be used
image_directory = "images_firstP"
image_name = "ezgif-frame-"

output_path = "output_masque_" +image_directory
```

```python
b_color_interval = [110, 200]
g_color_interval = [110, 190]
r_color_interval = [110, 190]

minim_pixels = 200000
```

```python
def check_path(path):
    if not os.path.exists(path):
        os.makedirs(path)
```

```python
def segment_image(image, b_color_interval, g_color_interval, r_color_interval):
    # Définition des bornes inférieures et supérieures en BGR
    lower_bound = np.array([b_color_interval[0], g_color_interval[0], r_color_interval[0]])
    upper_bound = np.array([b_color_interval[1], g_color_interval[1], r_color_interval[1]])

    mask = cv2.inRange(image, lower_bound, upper_bound)

    masked_image = cv2.bitwise_and(image, image, mask=mask)

    return mask, masked_image
```

```python
def largest_contour_aggreg_min(mask, minim_pixels):
    contours, _ = cv2.findContours(mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)


    all_large_contours = []
    for contour in contours:
        if cv2.contourArea(contour) > minim_pixels:
            all_large_contours.append(contour)
            #print(f"Contour area: {cv2.contourArea(contour)}")

    if len(all_large_contours) > 0:
        largest_contour = np.concatenate(all_large_contours)
    else :
        largest_contour = max(contours, key=cv2.contourArea)

    largest_mask = np.zeros_like(mask)

    cv2.drawContours(largest_mask, [largest_contour], -1, (255), thickness=cv2.FILLED)

    return largest_mask
```

```python
def find_top_pixel(mask):
    for i in range(mask.shape[0]):
        for j in range(mask.shape[1]):
            if mask[i, j] != 0:
                return j, i
```

```python
def cross_mark(image, top_p):
    cv2.drawMarker(image, top_p, (0, 0, 255), cv2.MARKER_CROSS, 20, 2)
    return image
```

```python
input_file = []

if not interval:
    input_file = os.listdir(image_directory)
else :
    for i in range(interval[0], interval[1]+1):
        if i < 10:
            input_file.append(f"{image_name}00{i}.png")
        elif i < 100:
            input_file.append(f"{image_name}0{i}.png")
        else:
            input_file.append(f"{image_name}{i}.png")
```

```python
for input in input_file:
    path = os.path.join(image_directory, input)
    print(path)
    image = cv2.imread(path)

    masque, segmented_image = segment_image(image, b_color_interval, g_color_interval, r_color_interval)

    largest_mask = largest_contour_aggreg_min(masque, minim_pixels)

    top_p = find_top_pixel(largest_mask)

    result = cv2.bitwise_and(image, image, mask=largest_mask)
    result = cross_mark(result, top_p)

    check_path(output_path)

    output_img = os.path.join(output_path, f"mask_{input}")
    print(output_img)

    cv2.imwrite(output_img, result)
```
